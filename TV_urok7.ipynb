```python
 #Вероятности для каждой нейросети
p_pastyr = 0.5  # вероятность, что нейросеть «Пастырь» распознает лицо
p_oko = 0.8     # вероятность, что нейросеть «Всевидящее око» распознает лицо
p_brat = 0.9    # вероятность, что нейросеть «Большой брат» распознает лицо

# Расчет вероятности для комбинации двух сетей
# Сначала работает «Большой брат», если он не распознает, задача передается «Всевидящему оку»
p_brat_oko = p_brat + (1 - p_brat) * p_oko

# Расчет вероятности для комбинации трех сетей
# Сначала работает «Большой брат», затем «Всевидящее око», и если обе не справились, задача передается «Пастырю»
p_brat_oko_pastyr = p_brat_oko + (1 - p_brat_oko) * p_pastyr

# Вывод результатов
print(f"Комбинация нейросетей 'Большой брат' и 'Всевидящее око': {p_brat_oko:.2f}")
print(f"Комбинация всех трех нейросетей: {p_brat_oko_pastyr:.2f}")

# Выбор нейросети или комбинации нейросетей с наибольшей вероятностью успеха
best_combination = max(p_brat, p_brat_oko, p_brat_oko_pastyr)
print(f"Наиболее эффективная комбинация с вероятностью успеха: {best_combination:.2f}")
```

```python
models = {
    "Всевидящее око": 0.8,
    "Большой брат": 0.9,
    "Пастырь": 0.5,
}

def chain_success(probs):
    """
    probs — список вероятностей моделей в порядке работы.
    Возвращает суммарную вероятность успеха цепочки.
    """
    p_fail_all_before = 1.0
    total_success = 0.0

    for p in probs:
        total_success += p_fail_all_before * p
        p_fail_all_before *= (1 - p)

    return total_success


# 1) Отдельные модели
print("Большой брат:", chain_success([models["Большой брат"]]))
print("Всевидящее око:", chain_success([models["Всевидящее око"]]))
print("Пастырь:", chain_success([models["Пастырь"]]))

# 2) Комбинация: Большой брат -> Всевидящее око
print("BB -> Око:", chain_success([
    models["Большой брат"],
    models["Всевидящее око"],
]))

# 3) Комбинация: Большой брат -> Всевидящее око -> Пастырь
print("BB -> Око -> Пастырь:", chain_success([
    models["Большой брат"],
    models["Всевидящее око"],
    models["Пастырь"],
]))
```
