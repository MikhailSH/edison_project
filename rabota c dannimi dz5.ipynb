```python
import pandas as pd
import matplotlib.pyplot as plt

from scipy.stats import pearsonr


# Загрузка данных
data = pd.read_csv("data111.csv")

# Просмотр данных
print("Первые строки:")
print(data.head())

print("\nТипы данных:")
print(data.dtypes)

print("\nПропущенные значения:")
print(data[["Price", "Value"]].isnull().sum())

# Удаление пропусков
analysis_data = data[
    ["Price", "Value"]
].dropna()

# Выделение переменных
price = analysis_data["Price"]
value = analysis_data["Value"]

# Расчёт корреляции Пирсона
correlation, p_value = pearsonr(price, value)

print("\nРезультаты анализа:")
print(f"Коэффициент корреляции: {correlation:.4f}")
print(f"p-value: {p_value:.4f}")

# Интерпретация статистической значимости
if p_value < 0.05:
    print("Корреляция статистически значима.")
else:
    print("Корреляция статистически незначима.")

# Интерпретация направления связи
if correlation > 0:
    print("Связь положительная.")
elif correlation < 0:
    print("Связь отрицательная.")
else:
    print("Линейная связь отсутствует.")

# Диаграмма рассеяния
plt.figure(figsize=(8, 6))

plt.scatter(
    price,
    value,
    alpha=0.8,
    edgecolors="black"
)

plt.xlabel("Цена")
plt.ylabel("Объём продаж")
plt.title("Зависимость объёма продаж от цены")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

from scipy.stats import pearsonr


# Загрузка данных
data = pd.read_csv("data111.csv")

# Проверка столбцов
print("Доступные столбцы:")
print(data.columns.tolist())

required_columns = ["Price", "Value"]

for column in required_columns:
    if column not in data.columns:
        raise KeyError(
            f"Столбец '{column}' не найден. "
            f"Доступные столбцы: {data.columns.tolist()}"
        )

# Преобразование в числовой формат
data["Price"] = pd.to_numeric(
    data["Price"],
    errors="coerce"
)

data["Value"] = pd.to_numeric(
    data["Value"],
    errors="coerce"
)

# Удаление строк с пропусками
analysis_data = data[
    ["Price", "Value"]
].dropna()

if len(analysis_data) < 3:
    raise ValueError(
        "Для расчёта корреляции нужно минимум 3 строки."
    )

price = analysis_data["Price"]
value = analysis_data["Value"]

# Расчёт корреляции
correlation, p_value = pearsonr(price, value)

print("\nРезультаты анализа:")
print(f"Коэффициент корреляции: {correlation:.4f}")
print(f"p-value: {p_value:.4f}")

# Интерпретация
if p_value < 0.05:
    print("Корреляция статистически значима.")
else:
    print("Корреляция статистически незначима.")

if correlation > 0:
    print("Связь положительная.")
elif correlation < 0:
    print("Связь отрицательная.")
else:
    print("Линейная связь отсутствует.")

# Диаграмма рассеяния
plt.figure(figsize=(8, 6))

plt.scatter(
    price,
    value,
    alpha=0.7,
    edgecolors="black"
)

plt.xlabel("Цена")
plt.ylabel("Объём продаж")
plt.title("Зависимость объёма продаж от цены")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```
