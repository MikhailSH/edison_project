```python
class Database:
    def __init__(self, host, port, database, user, password, autocomit=False):
        self.connection = psycopg2.connect(
            host = host,
            port = port,
            database = database,
            user = user,
            password = password,
        )
        if autocomit:
            self.connection.autocommit = True

        self.cursor = self.connection.cursor()
    def select(self, query, vars):
        self.cursor.execute(query, vars)
        res = self.cursor.fetchall()
        return res
    def post(self, query, vars):
        self.cursor.execute(query, vars)
        if not self.connection.autocommit:
            self.connection.commit()
```

```python
from psycopg2.extras import RealDictCursor
```

```python
class Database:
    def __init__(self, host, port, database, user, password, autocomit=False):
        self.connection = psycopg2.connect(
            host=host,
            port=port,
            database=database,
            user=user,
            password=password,
        )
        if autocomit:
            self.connection.autocommit = True

        self.cursor = self.connection.cursor(cursor_factory=RealDictCursor)

    def select(self, query, vars):
        self.cursor.execute(query, vars)
        res = self.cursor.fetchall()
        return res

    def post(self, query, vars):
        self.cursor.execute(query, vars)
        if not self.connection.autocommit:
            self.connection.commit()
```

```python
db = Database(HOST, PORT, DATABASE, USER, PASSWORD)
```

```python
res = db.select("select * from order_details where quantity > %s limit 5", (1, ))
```
