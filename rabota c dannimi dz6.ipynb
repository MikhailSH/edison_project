```python
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import pearsonr

data = pd.read_csv("Customer Segmentation.csv")

price = data["Price"]
value = data["Value"]

correlation, p_value = pearsonr(price, value)

print(f"Корреляция: {correlation:.4f}")
print(f"p-value: {p_value:.4f}")

plt.scatter(price, value)
plt.xlabel("Price")
plt.ylabel("Value")
plt.title("Зависимость Value от Price")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import pearsonr

data = pd.read_csv("Customer Segmentation.csv")

analysis_data = data[
    ["Age", "Family_Size"]
].dropna()

correlation, p_value = pearsonr(
    analysis_data["Age"],
    analysis_data["Family_Size"]
)

print(f"Корреляция: {correlation:.4f}")
print(f"p-value: {p_value:.4f}")

plt.scatter(
    analysis_data["Age"],
    analysis_data["Family_Size"]
)

plt.xlabel("Возраст")
plt.ylabel("Размер семьи")
plt.title("Связь возраста и размера семьи")
plt.show()
```
